# 復習ドリル🐼  Python練習版
このノートブックでは、**Pythonでランダムに選んだ3語から1語をHTML表示し、ボタンで日本語訳を出す**練習をします。

1. 最初のセルを実行して練習用単語を準備
2. 本物の復習帳を試したい場合だけ、2番目のセルでChromeから書き出した `review_book_export.json` をアップロード（省略OK）
3. 最後のセルを実行し、「答えを見る🐼」をクリック

**注意：** ここでの表示はGoogle Colab内だけです。GitHub PagesのTrainer画面は変更せず、復習帳からの単語削除も行いません。個人メモを含むJSONは公開GitHubにアップロードしないでください。

```python
import random
import ipywidgets as widgets
from IPython.display import HTML, display

words = [
    {"thai": "กิน", "japanese": "食べる"},
    {"thai": "ไป", "japanese": "行く"},
    {"thai": "มา", "japanese": "来る"}
]
```

## （任意）実際の復習帳を読み込む
Trainerの「復習帳をJSON保存🐍」でダウンロードしたファイルを使うなら、**このセルを実行**してください。練習用の3語だけを使う場合は飛ばしてOKじゃ。

```python
# 本物の復習帳を使う場合だけ実行する
import json
from google.colab import files

uploaded = files.upload()
if uploaded:
    file_bytes = next(iter(uploaded.values()))
    review_book = json.loads(file_bytes.decode("utf-8-sig"))
    if not isinstance(review_book, list):
        raise ValueError("復習帳のJSONはリスト形式にしてください")
    words = [word for word in review_book if isinstance(word, dict) and word.get("thai")]
    print(f"復習帳から{len(words)}語を読み込みました🐍")
```

## Pythonで出題し、ボタンを押すと答えをHTML表示
`random.sample()`、`len()`、`HTML()`、`widgets.Button()`、`on_click()` を使います。

```python
if not words:
    display(HTML("<p>出題できる単語がありません。</p>"))
else:
    selected_words = random.sample(words, k=min(3, len(words)))
    i = 0

    html = "<h2>復習ドリル🐼</h2>"
    html += f"<p>{i+1}問目/{len(selected_words)}問</p>"
    html += f"<p>{selected_words[i]['thai']}</p>"
    display(HTML(html))

    button = widgets.Button(description="答えを見る 🐼")
    answer_area = widgets.Output()

    def show_answer(b):
        with answer_area:
            answer_area.clear_output(wait=True)
            html = f"<h2>正解：{selected_words[i].get('japanese') or '未登録'}</h2>"
            display(HTML(html))

    button.on_click(show_answer)
    display(button, answer_area)
```
